### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5
Preço do litro (R$): 5.89
Total de pedágios (R$): 32.40
Número de pessoas: 3

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
def read_float(prompt):
    """Lê um número decimal do teclado aceitando vírgula ou ponto.

    prompt: str exibida para o usuário
    Retorna o número digitado (float).
    """
    typed_value = input(prompt)
    # o Python aceita . Pra que funcione com ','
    return float(typed_value.replace(",", "."))


distance_km = read_float("Distância da viagem (km): ")
fuel_efficiency = read_float("Consumo médio (km/l): ")
fuel_price = read_float("Preço do litro (R$): ")
toll_total = read_float("Total de pedágios (R$): ")
people_count = int(input("Número de pessoas: "))

liters_needed = distance_km / fuel_efficiency
fuel_cost = liters_needed * fuel_price
total_cost = fuel_cost + toll_total
cost_per_person = total_cost / people_count

print()
print(f"Litros necessários: {round(liters_needed, 2)}")
print(f"Custo do combustível: R$ {fuel_cost:.2f}")
print(f"Custo total da viagem: R$ {total_cost:.2f}")
print(f"Custo por pessoa: R$ {cost_per_person:.2f}")
print(f"Tipo da distância: {type(distance_km)} | Tipo do número de pessoas: {type(people_count)}")
````

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;

### Questão 6: Desenhando com `jupyturtle` (1,0 ponto)

**(a)** Escreva uma função chamada `poligono(lados, tamanho)` que desenhe um polígono regular com o número de lados e o comprimento de lado informados. A função deve ter docstring.

**(b)** Escreva uma função chamada `flor(petalas, lados, tamanho)` que **reutilize** `poligono` para desenhar
uma flor.

**(c)** Faça três desenhos: um triângulo, um hexágono e uma flor.

Exemplo:

```

flor(8, 4, 30)
```

````python
from os.path import basename, exists


def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename


download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py');
from jupyturtle import make_turtle, forward, left
````

````python
def polygon(sides, size):
    """Draws a regular polygon.

    sides: number of sides of the polygon (int)
    size: length of each side, in turtle steps
    """
    turn_angle = 360 / sides
    for i in range(sides):
        forward(size)
        left(turn_angle)


````

````python
def flower(petals, sides, size):
    """Draws a flower by reusing the polygon function.

    petals: number of petals of the flower (int)
    sides: number of sides of each petal
    size: length of the side of each petal

    Each petal is a polygon; between one petal and the next the turtle
    turns 360 / petals degrees, spreading the petals around the center.
    """
    angle_between_petals = 360 / petals
    for i in range(petals):
        polygon(sides, size)
        left(angle_between_petals)
````

````python
# (c) triangle
make_turtle()
polygon(3, 80)

# (c) hexagon
make_turtle()
polygon(6, 30)

# (c) flower
make_turtle()
flower(8, 4, 30)
````

**(b)** Escreva uma função chamada `eh_triangulo(a, b, c)` que retorne `True` se três medidas podem formar um
triângulo. A condição é que cada lado seja **positivo** e que cada lado seja **menor** que a soma dos outros dois.

**(c)** Usando `for i in range(...)`, exiba para os anos de **1996 a 2005** se cada um é bissexto ou não.

**(d)** Teste `eh_triangulo` com pelo menos três combinações diferentes, exibindo o resultado de cada teste.

Exemplo:

```

1997 é bissexto? False
...
2000 é bissexto? True
...
eh_triangulo(3, 4, 5) -> True
eh_triangulo(1, 2, 10) -> False
eh_triangulo(-1, 4, 5) -> False
```

````python
def is_leap_year(year):
    """Tells whether a year is a leap year.

    year: the year to be tested (int)

    Returns True if the year is a leap year, False otherwise.
    Rule: it is a leap year when divisible by 4 and not by 100,
    or when divisible by 400.
    """
    return year % 4 == 0 and (not year % 100 == 0 or year % 400 == 0)


def is_triangle(side_a, side_b, side_c):
    """Tells whether three measures can form a triangle.

    side_a: measure of the first side (float)
    side_b: measure of the second side (float)
    side_c: measure of the third side (float)

    Returns True when all sides are positive and each side is shorter
    than the sum of the other two; otherwise returns False.
    """
    all_sides_positive = side_a > 0 and side_b > 0 and side_c > 0
    triangle_inequality = (side_a < side_b + side_c and
                           side_b < side_a + side_c and
                           side_c < side_a + side_b)
    return all_sides_positive and triangle_inequality


for year in range(1996, 2006):
    print(f"Is {year} a leap year? {is_leap_year(year)}")

triangle_tests = {
    "is_triangle(3, 4, 5)": is_triangle(3, 4, 5),
    "is_triangle(1, 2, 10)": is_triangle(1, 2, 10),
    "is_triangle(-1, 4, 5)": is_triangle(-1, 4, 5),
    "is_triangle(6, 6, 6)": is_triangle(6, 6, 6),
}

print()
for call_text in triangle_tests:
    print(f"{call_text} -> {triangle_tests[call_text]}")
````

````python
status_rules = {
    "Approved": (7, float("inf")),
    "Recovery": (4, 7),
    "Failed": (float("-inf"), 4),
}


def average_grades(grade1, grade2, grade3):
    """Computes the arithmetic mean of three grades.

    grade1: first grade (float)
    grade2: second grade (float)
    grade3: third grade (float)

    Returns the average of the three grades (float).
    """
    return (grade1 + grade2 + grade3) / 3


def student_status(average):
    """Tells the status of a student from the average.

    average: the average of the student (float)

    Returns "Approved" (average 7 or higher), "Recovery" (from 4 to
    below 7) or "Failed" (below 4). The ranges in status_rules cover
    every real number, so there is always a match.
    """
    for status in status_rules:
        lower_bound = status_rules[status][0]
        upper_bound = status_rules[status][1]
        if lower_bound <= average < upper_bound:
            return status


sample_grades = {
    "Ana": (7, 5.5, 9),
    "Bruno": (4, 3, 2),
    "Carla": (9.5, 8, 10),
}

for student_name in sample_grades:
    grades = sample_grades[student_name]
    student_average = average_grades(grades[0], grades[1], grades[2])
    print(f"{student_name}: {student_average:.2f} -> {student_status(student_average)}")

# tests
print(student_status(7.0))   # expected: Approved
print(student_status(6.99))  # expected: Recovery
print(student_status(4.0))   # expected: Recovery
print(student_status(3.9))   # expected: Failed
print(student_status(0))     # expected: Failed
````
